# Extracting data from page 1-10

```python
#Extracting data from page 1-10

base_url = "https://quotes.toscrape.com/page/{}/"


```

```python
all_quotes = []
all_authors = []
all_tags = []
```

```python
for page in range(1, 11):
    #page 1 - 10
    url = base_url.format(page)
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    quotes = soup.find_all("div", class_ = "quote")
    
    for quote in quotes:
        text = quote.find("span", class_="text").text
        author = quote.find("small", class_="author").get_text()
        tags = [tag.get_text() for tag in quote.find_all("a", class_="tag")]
        
        all_quotes.append(text)
        all_authors.append(author)
        all_tags.append(','.join(tags))
        
        
print(f"found {len(all_quotes)} quotes total\n")
        
        
```

```python
#Saving the extracted quotes, author and tags into csv file

import csv
```

```python
with open("quotes.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Quotes", "Author", "Tags"])
    
    for text, author, tags in zip(all_quotes, all_authors, all_tags):
        writer.writerow([text, author, tags])
        
print(f"Done {len(all_quotes)} quotes saved to quotes.csv")
```
